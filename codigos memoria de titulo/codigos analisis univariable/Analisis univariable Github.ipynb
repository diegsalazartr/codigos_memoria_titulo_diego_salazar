## 1) Extracciones de cobre

### 1.2) Botplot de extracciones de cobre

```python
import pandas as pd
import matplotlib.pyplot as plt

excelPath = r"C:\Users\Diego\Desktop\Copia de Base de Datos Cuprochlor Sin T.xlsx"

# Leer Excel
df = pd.read_excel(excelPath, sheet_name="BD elim y reducido (2)")
df.columns = df.columns.str.strip()

# Orden solicitado
columnas_ordenadas = [
    'ext_cut_anlz_ph',         # Iso pH – Análisis de solución
    'ext_cut_anlz_col',        # Columna – Análisis de solución
    'ext_cut_rip_ph',          # Iso pH – Ripios
    'ext_cut_rip_selec_col'    # Columna – Ripios
]

labels_personalizados = [
    'Iso pH – Cabeza Analizada',
    'Columna – Cabeza Analizada',
    'Iso pH – Tierras',
    'Columna – Tierras'
]

# Validación
faltantes = [c for c in columnas_ordenadas if c not in df.columns]
if faltantes:
    raise ValueError(f"Faltan estas columnas: {faltantes}")

# Datos
data_boxplot = [df[c].dropna().values for c in columnas_ordenadas]

# Figura
plt.figure(figsize=(9, 5))
plt.boxplot(data_boxplot, labels=labels_personalizados)
plt.ylabel('Extracción de cobre [%]')
plt.xticks(rotation=30, ha='right')
plt.tight_layout()

# GUARDAR FIGURA (TESIS)
plt.savefig(r"C:\Users\Diego\Desktop\Boxplot_extraccion_cobre.png", dpi=300)

plt.show()
```

## 5) Distribución mineralógica

```python
import pandas as pd
import matplotlib.pyplot as plt

# --- CONFIGURACIÓN ---
nombre_archivo = "Distribucion_Muestras_por_Zona.png" # <--- Nombre para guardar
# ---------------------

# 1. Carga de datos (Solo necesitamos la columna de Zona Min)
columnas_deseadas = ['Zona Min']
excelPath = r"C:\Users\Diego\Desktop\Ultimo Esfuerzo\Copia de Base de Datos Cuprochlor Sin T Final.xlsx"

# Cargamos y limpiamos valores nulos
df = pd.read_excel(excelPath, "BD elim y reducido (2)", usecols=columnas_deseadas).dropna()

# 2. Contar la cantidad de muestras por cada zona
conteo_zonas = df['Zona Min'].value_counts()

# 3. Creación del gráfico de barras
plt.figure(figsize=(10, 6))

# Definimos colores para cada zona (Naranja para Óxidos, Verde para Secundarios, Salmón para Primarios)
colores = ['skyblue', 'skyblue', 'skyblue']

# Creamos las barras
# conteo_zonas.index contiene los nombres (CUO, SSE, PRI)
# conteo_zonas.values contiene las cantidades
barras = plt.bar(conteo_zonas.index, conteo_zonas.values, color=colores, edgecolor='black', alpha=0.8)

# 4. Personalización estética
plt.title('', fontsize=14, fontweight='bold')
plt.xlabel('Zona Mineralógica', fontsize=12)
plt.ylabel('Cantidad de Muestras', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Añadir el número exacto sobre cada barra para mayor claridad
for barra in barras:
    yval = barra.get_height()
    plt.text(barra.get_x() + barra.get_width()/2, yval + 1, int(yval), 
             ha='center', va='bottom', fontweight='bold')

# 5. Guardar y mostrar
plt.savefig(nombre_archivo, dpi=300, bbox_inches='tight')
print(f"Gráfico guardado como: {nombre_archivo}")
print("Conteo por zona:")
print(conteo_zonas)

plt.show()
```
