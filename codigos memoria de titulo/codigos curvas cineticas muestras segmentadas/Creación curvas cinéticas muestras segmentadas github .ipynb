```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ============================================================
# CONFIGURACIÓN EDITABLE
# ============================================================

# ------------------------------------------------------------
# ARCHIVO EXCEL
# ------------------------------------------------------------

archivo_excel = r"C:\Users\Diego\calculo graficas 3.xlsx"


# ============================================================
# CONFIGURACIÓN DE NOMBRES
# ============================================================

# Nombres de las curvas
nombre_curva_iso = "Iso pH cabeza analizada"
nombre_curva_col = "Columna cabeza analizada"

# Nombres de los puntos
nombre_punto_iso_rip = "Iso pH tierra"
nombre_punto_col_rip = "Columna tierras"

# Nombres de los ejes
nombre_eje_x = "Tiempo normalizado"
nombre_eje_y = "Extracción [%]"


# ============================================================
# CONFIGURACIÓN VISUAL
# ============================================================

# Tamaño de la figura
tamaño_figura = (10, 6)

# Tamaño de las letras de los ejes
tamaño_fuente_ejes = 14

# Tamaño de las letras de la leyenda
tamaño_fuente_leyenda = 13

# Tamaño de los marcadores de las curvas
tamaño_marcador_curvas = 5

# Tamaño de los puntos Iso rip y Col rip
tamaño_puntos_rip = 100


# ============================================================
# CONFIGURACIÓN DE LA CARPETA DE SALIDA
# ============================================================

# Crear carpeta "graficas segmentadas"
carpeta_salida = Path("graficas segmentadas")

# Si la carpeta no existe, se crea automáticamente
carpeta_salida.mkdir(
    parents=True,
    exist_ok=True
)


# ============================================================
# OBTENER TODAS LAS HOJAS DEL ARCHIVO EXCEL
# ============================================================

# Leer los nombres de todas las hojas
todas_las_hojas = pd.ExcelFile(
    archivo_excel
).sheet_names


# ============================================================
# FILTRAR SOLAMENTE LAS HOJAS GMZV-X
# ============================================================

hojas_gmzv = [
    hoja
    for hoja in todas_las_hojas
    if hoja.upper().startswith("GMZV-")
]


# ============================================================
# MOSTRAR LAS HOJAS QUE SERÁN PROCESADAS
# ============================================================

print("=" * 60)
print("HOJAS GMZV ENCONTRADAS")
print("=" * 60)

for hoja in hojas_gmzv:
    print(f"  - {hoja}")

print(
    f"\nTotal de hojas a procesar: "
    f"{len(hojas_gmzv)}"
)

print("=" * 60)


# ============================================================
# PROCESAR CADA HOJA
# ============================================================

for hoja in hojas_gmzv:

    print("\n" + "=" * 60)
    print(f"PROCESANDO: {hoja}")
    print("=" * 60)

    try:

        # ----------------------------------------------------
        # LEER LA HOJA ACTUAL
        # ----------------------------------------------------

        df = pd.read_excel(
            archivo_excel,
            sheet_name=hoja
        )


        # ----------------------------------------------------
        # VERIFICAR QUE EXISTAN LAS COLUMNAS NECESARIAS
        # ----------------------------------------------------

        columnas_necesarias = [
            "tiempo n iso",
            "ext iso anlz",
            "tiempo n col",
            "ext col anlz",
            "iso rip",
            "col rip"
        ]

        columnas_faltantes = [
            columna
            for columna in columnas_necesarias
            if columna not in df.columns
        ]

        if columnas_faltantes:

            print(
                f"⚠️ Se omitió {hoja}."
            )

            print(
                "Columnas faltantes:",
                columnas_faltantes
            )

            continue


        # ====================================================
        # DATOS DE LA CURVA ISO pH
        # ====================================================

        iso = df[
            [
                "tiempo n iso",
                "ext iso anlz"
            ]
        ].dropna()


        # ====================================================
        # DATOS DE LA CURVA COLUMNA
        # ====================================================

        col = df[
            [
                "tiempo n col",
                "ext col anlz"
            ]
        ].dropna()


        # ====================================================
        # DATOS ISO RIP
        # ====================================================

        iso_rip = df[
            [
                "tiempo n iso",
                "iso rip"
            ]
        ].dropna()


        # ====================================================
        # DATOS COL RIP
        # ====================================================

        col_rip = df[
            [
                "tiempo n col",
                "col rip"
            ]
        ].dropna()


        # ====================================================
        # VERIFICAR QUE EXISTAN DATOS SUFICIENTES
        # ====================================================

        if iso.empty:

            print(
                f"⚠️ {hoja}: "
                f"no hay datos para la curva Iso pH."
            )

            continue


        if col.empty:

            print(
                f"⚠️ {hoja}: "
                f"no hay datos para la curva Columna."
            )

            continue


        if iso_rip.empty:

            print(
                f"⚠️ {hoja}: "
                f"no hay datos para Iso rip."
            )

            continue


        if col_rip.empty:

            print(
                f"⚠️ {hoja}: "
                f"no hay datos para Col rip."
            )

            continue


        # ====================================================
        # OBTENER ÚLTIMO VALOR DE TIEMPO Y EXTRACCIÓN
        # ====================================================

        # ----------------------------------------------------
        # PUNTO ISO RIP
        # ----------------------------------------------------

        x_iso_rip = iso_rip[
            "tiempo n iso"
        ].iloc[-1]

        y_iso_rip = iso_rip[
            "iso rip"
        ].iloc[-1]


        # ----------------------------------------------------
        # PUNTO COL RIP
        # ----------------------------------------------------

        x_col_rip = col_rip[
            "tiempo n col"
        ].iloc[-1]

        y_col_rip = col_rip[
            "col rip"
        ].iloc[-1]


        # ====================================================
        # CREAR GRÁFICA
        # ====================================================

        plt.figure(
            figsize=tamaño_figura
        )


        # ====================================================
        # CURVA ISO pH
        # ====================================================

        plt.plot(
            iso["tiempo n iso"],
            iso["ext iso anlz"],
            marker="o",
            markersize=tamaño_marcador_curvas,
            linewidth=2,
            label=nombre_curva_iso
        )


        # ====================================================
        # CURVA COLUMNA
        # ====================================================

        plt.plot(
            col["tiempo n col"],
            col["ext col anlz"],
            marker="o",
            markersize=tamaño_marcador_curvas,
            linewidth=2,
            label=nombre_curva_col
        )


        # ====================================================
        # PUNTO ISO RIP
        # ====================================================

        plt.scatter(
            x_iso_rip,
            y_iso_rip,
            s=tamaño_puntos_rip,
            marker="D",
            label=nombre_punto_iso_rip,
            zorder=5
        )


        # ====================================================
        # PUNTO COL RIP
        # ====================================================

        plt.scatter(
            x_col_rip,
            y_col_rip,
            s=tamaño_puntos_rip,
            marker="D",
            label=nombre_punto_col_rip,
            zorder=5
        )


        # ====================================================
        # FORMATO DE LA GRÁFICA
        # ====================================================

        # NO SE AGREGA TÍTULO
        # El título se incorporará posteriormente en la tesis

        # Eje X
        plt.xlabel(
            nombre_eje_x,
            fontsize=tamaño_fuente_ejes
        )

        # Eje Y
        plt.ylabel(
            nombre_eje_y,
            fontsize=tamaño_fuente_ejes
        )

        # Leyenda
        plt.legend(
            fontsize=tamaño_fuente_leyenda
        )

        # Cuadrícula
        plt.grid(
            True,
            alpha=0.3
        )

        # Ajustar automáticamente los márgenes
        plt.tight_layout()


        # ====================================================
        # GUARDAR GRÁFICA
        # ====================================================

        nombre_archivo = (
            f"grafica_{hoja}.png"
        )

        ruta_salida = (
            carpeta_salida /
            nombre_archivo
        )


        plt.savefig(
            ruta_salida,
            dpi=300,
            bbox_inches="tight"
        )


        # ====================================================
        # CERRAR FIGURA
        # ====================================================

        # Importante para no acumular muchas figuras
        # en memoria al procesar todas las hojas
        plt.close()


        # ====================================================
        # MOSTRAR INFORMACIÓN
        # ====================================================

        print(
            f"✓ Gráfica creada: "
            f"{nombre_archivo}"
        )

        print(
            f"  Iso rip -> "
            f"Tiempo = {x_iso_rip:.6f}, "
            f"Extracción = {y_iso_rip:.6f}"
        )

        print(
            f"  Col rip -> "
            f"Tiempo = {x_col_rip:.6f}, "
            f"Extracción = {y_col_rip:.6f}"
        )


    # ========================================================
    # CONTROL DE ERRORES
    # ========================================================

    except Exception as e:

        print(
            f"❌ Error procesando {hoja}:"
        )

        print(
            f"   {e}"
        )

        continue


# ============================================================
# FINALIZACIÓN
# ============================================================

print("\n" + "=" * 60)
print("PROCESO FINALIZADO")
print("=" * 60)

print(
    "Las gráficas fueron guardadas en:"
)

print(
    carpeta_salida.resolve()
)

print("=" * 60)
```
