# Test de hipótesis para sulfuros secundarios de cobre segmentado por control de calidad

## 1) Test de hipótesis para extracción de cobre mediante tierras

### 1.1) Test de hipotesis normalidad para sulfuros secundarios de cobre metodo de extraccion por tierras

```python
# 1. Definición de columnas y carga de datos
# En las columnas deseadas, Zona Min como nueva columna para ser utilizada como filtro
columnas_deseadas = ['Zona Min', 'ext_cut_rip_ph', 'ext_cut_rip_selec_col']
excelPath = r"C:\Users\Diego\Desktop\Ultimo Esfuerzo\Copia de Base de Datos Cuprochlor Sin T Final.xlsx"

df3 = pd.read_excel(excelPath, "BD elim y reducido (3)", usecols=columnas_deseadas)

# 2. Cálculo de la variable diferencia
df3['diferencia'] = df3['ext_cut_rip_ph'] - df3['ext_cut_rip_selec_col']

# 3. Filtrado por zona mineralógica (Óxidos de Cobre - CUO)
# Limpiamos valores nulos específicos de este segmento
datos_oxidos = df3[df3['Zona Min'] == 'SSE']['diferencia'].dropna()

# 4.Calculo de asimetría 
datos=datos_oxidos
asimetria=skew(datos)
print("Asimetria:",asimetria)

# 5.Calculo de Curtosis
datos=datos_oxidos
curtosis=kurtosis(datos)
print("Curtosis:",curtosis)

# 6. Realizar el Test de Shapiro-Wilk
# Nota: Este test no requiere parámetros adicionales (mu, sigma)
shapiro_stat, p_value = stats.shapiro(datos_oxidos)

# 7. Mostrar e interpretar resultados
print("--- TEST DE NORMALIDAD (SHAPIRO-WILK) - ZONA: SSE (SULFUROS SECUNDARIOS) ---")
print(f"Estadístico W: {shapiro_stat:.4f}")
print(f"P-valor: {p_value:.4f}")

print("\nInterpretación:")
alpha = 0.05
if p_value > alpha:
    print("Resultado: Los datos de la zona CUO parecen seguir una distribución NORMAL.")
    print("Interpretación: No se rechaza la hipótesis nula (p > 0.05).")
else:
    print("Resultado: Los datos de la zona CUO NO siguen una distribución normal.")
    print("Interpretación: Se rechaza la hipótesis nula (p < 0.05).")

print(f"\nCantidad de muestras analizadas en CUO: {len(datos_oxidos)}")
```

### 1.2) Test t Student para sulfuros secundarios de cobre metodo de extraccion por tierras

```python
import pandas as pd
from scipy import stats

# 1. Definición de columnas y carga de datos
columnas_deseadas = ['Zona Min', 'ext_cut_rip_ph', 'ext_cut_rip_selec_col']
excelPath = r"C:\Users\Diego\Desktop\Ultimo Esfuerzo\Copia de Base de Datos Cuprochlor Sin T Final.xlsx"

df3 = pd.read_excel(excelPath, "BD elim y reducido (3)", usecols=columnas_deseadas)

# 2. Filtrado por zona (SSE - Sulfuros Secundarios) y limpieza de nulos
# Aseguramos que ambas columnas tengan datos para realizar la comparación por pares
df_sse = df3[df3['Zona Min'] == 'SSE'].dropna(subset=['ext_cut_rip_ph', 'ext_cut_rip_selec_col'])

# 3. Realizar el Test T de Student para muestras relacionadas (Paired T-Test)
t_stat, p_value = stats.ttest_rel(df_sse['ext_cut_rip_ph'], df_sse['ext_cut_rip_selec_col'])

# 4. Mostrar e interpretar resultados
print("--- TEST T DE STUDENT (MUESTRAS RELACIONADAS) - ZONA: SSE (SULFUROS SECUNDARIOS) ---")
print(f"Estadístico t: {t_stat:.4f}")
print(f"P-valor: {p_value:.10f}")
print(f"Cantidad de muestras analizadas en SSE: {len(df_sse)}")

print("\nInterpretación:")
alpha = 0.05
if p_value < alpha:
    print("Resultado: SIGNIFICATIVO.")
    print("En la zona de Sulfuros Secundarios (SSE), existe una diferencia estadística real entre los métodos.")
    
    # Comparación de medias para determinar eficiencia en SSE
    media_ph = df_sse['ext_cut_rip_ph'].mean()
    media_col = df_sse['ext_cut_rip_selec_col'].mean()
    
    if media_ph > media_col:
        print(f"La extracción Iso pH ({media_ph:.2f}%) es significativamente superior a la de Columna ({media_col:.2f}%).")
    else:
        print(f"La extracción por Columna ({media_col:.2f}%) es significativamente superior a la de Iso pH ({media_ph:.2f}%).")
else:
    print("Resultado: NO SIGNIFICATIVO.")
    print("En la zona SSE, no hay evidencia suficiente para afirmar que un método sea diferente al otro (p > 0.05).")
```

## 2) Test de hipótesis para extracción de cobre mediante cabeza analizada

### 2.1) Test de hipotesis normalidad para sulfuros secundarios de cobre metodo cabeza analizada

```python
##### import pandas as pd
from scipy import stats
from scipy.stats import skew, kurtosis

# 1. Definición de columnas y carga de datos (Análisis de Solución)
columnas_deseadas = ['Zona Min', 'ext_cut_anlz_ph', 'ext_cut_anlz_col']
excelPath = r"C:\Users\Diego\Desktop\Ultimo Esfuerzo\Copia de Base de Datos Cuprochlor Sin T Final.xlsx"

# Cargamos el DataFrame df4
df4 = pd.read_excel(excelPath, "BD elim y reducido (3)", usecols=columnas_deseadas)

# 2. Cálculo de la variable diferencia (Análisis de Solución)
df4['diferencia'] = df4['ext_cut_anlz_ph'] - df4['ext_cut_anlz_col']

# 3. Filtrado por zona mineralógica (Sulfuros Secundarios - SSE)
# Limpiamos nulos específicos para la variable 'diferencia' en esta zona
datos_sse = df4[df4['Zona Min'] == 'SSE']['diferencia'].dropna()

# 4. Cálculo de Asimetría
asimetria = skew(datos_sse)
print(f"Asimetría (SSE - Solución): {asimetria:.4f}")

# 5. Cálculo de Curtosis
curto = kurtosis(datos_sse)
print(f"Curtosis (SSE - Solución): {curto:.4f}")

# 6. Parámetros para el Test Kolmogorov-Smirnov (Específicos para SSE)
mu_sse = datos_sse.mean()
sigma_sse = datos_sse.std()

# 7. Realizar el Test de Kolmogorov-Smirnov
# Comparamos la muestra de SSE contra una normal teórica ajustada
d_stat, p_value = stats.kstest(datos_sse, 'norm', args=(mu_sse, sigma_sse))

# 8. Mostrar e interpretar resultados
print("\n--- TEST DE NORMALIDAD (KOLMOGOROV-SMIRNOV) - ZONA: SSE (SULFUROS SECUNDARIOS) ---")
print(f"Estadístico D: {d_stat:.4f}")
print(f"P-valor: {p_value:.4f}")
print(f"Cantidad de muestras analizadas en SSE: {len(datos_sse)}")

print("\nInterpretación:")
alpha = 0.05
if p_value > alpha:
    print("Resultado: Los datos de Sulfuros Secundarios (SSE) parecen seguir una distribución NORMAL.")
    print("Interpretación: No se rechaza la hipótesis nula (p > 0.05).")
else:
    print("Resultado: Los datos de Sulfuros Secundarios (SSE) NO siguen una distribución normal.")
    print("Interpretación: Se rechaza la hipótesis nula (p < 0.05).")
```

### 2.2) Test t Student para la extracción de sulfuros secundarios de cobre calculada mediante método de cabeza analizada

```python
## import pandas as pd
from scipy import stats

# 1. Definición de columnas y carga de datos (Análisis de Solución)
columnas_deseadas = ['Zona Min', 'ext_cut_anlz_ph', 'ext_cut_anlz_col']
excelPath = r"C:\Users\Diego\Desktop\Ultimo Esfuerzo\Copia de Base de Datos Cuprochlor Sin T Final.xlsx"

# Cargamos el DataFrame df4
df4 = pd.read_excel(excelPath, "BD elim y reducido (3)", usecols=columnas_deseadas)

# 2. Filtrado por zona (SSE - Sulfuros Secundarios) y limpieza de nulos
# Filtramos primero la zona y luego aseguramos que existan ambos datos de solución
df_sse_anlz = df4[df4['Zona Min'] == 'SSE'].dropna(subset=['ext_cut_anlz_ph', 'ext_cut_anlz_col'])

# 3. Realizar el Test T de Student para muestras relacionadas (Paired T-Test)
t_stat, p_value = stats.ttest_rel(df_sse_anlz['ext_cut_anlz_ph'], df_sse_anlz['ext_cut_anlz_col'])

# 4. Mostrar e interpretar resultados
print("--- TEST T DE STUDENT (MUESTRAS RELACIONADAS) - ZONA: SSE (SULFUROS SECUNDARIOS - SOLUCIÓN) ---")
print(f"Estadístico t: {t_stat:.4f}")
print(f"P-valor: {p_value:.10f}")
print(f"Cantidad de muestras analizadas en SSE: {len(df_sse_anlz)}")

print("\nInterpretación:")
alpha = 0.05
if p_value < alpha:
    print("Resultado: SIGNIFICATIVO.")
    print("En la zona de Sulfuros Secundarios (SSE), existe una diferencia estadística real en la extracción vía solución.")
    
    # Comparación de medias dinámicas para el reporte
    media_ph = df_sse_anlz['ext_cut_anlz_ph'].mean()
    media_col = df_sse_anlz['ext_cut_anlz_col'].mean()
    
    if media_ph > media_col:
        print(f"La extracción Iso pH ({media_ph:.2f}%) es significativamente superior a la de Columna ({media_col:.2f}%).")
    else:
        print(f"La extracción por Columna ({media_col:.2f}%) es significativamente superior a la de Iso pH ({media_ph:.2f}%).")
else:
    print("Resultado: NO SIGNIFICATIVO.")
    print("En la zona SSE (solución), no hay evidencia suficiente para afirmar que un método sea diferente al otro (p > 0.05).")
```
