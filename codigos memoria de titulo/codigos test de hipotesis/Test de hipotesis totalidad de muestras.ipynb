```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import ttest_rel
import warnings
warnings.filterwarnings("ignore")
from scipy.stats import skew
from scipy.stats import kurtosis
from scipy.stats import wilcoxon
```

# Test de hipótesis para la totalidad de las muestras

## 1) Test de hipótesis para extracción de cobre mediante tierras

### 1.1) Test de hipotesis normalidad para la totalidad de las muestras metodo de extraccion por tierras

```python
# Definición de columnas y carga de datos.

columnas_deseadas=['ext_cut_rip_selec_col', 'ext_cut_rip_ph']
excelPath=r"C:\Users\Diego\Desktop\Ultimo Esfuerzo\Copia de Base de Datos Cuprochlor Sin T Final.xlsx"
dataframe=pd.read_excel(excelPath,"BD elim y reducido (2)",usecols=columnas_deseadas)
df3=dataframe

# 2. Cálculo de la variable diferencia
df3['diferencia']=df3['ext_cut_rip_ph']-df3['ext_cut_rip_selec_col']

#Calculo de asimetría 
datos=df3['diferencia']
asimetria=skew(datos)
print("Asimetria:",asimetria)

#Calculo de Curtosis
datos=df3['diferencia']
curtosis=kurtosis(datos)
print("Curtosis:",curtosis)

# 1. Limpieza de datos (Eliminar valores nulos para que el test no falle)
datos_diferencia = df3['diferencia'].dropna()

# 2. Calcular los parámetros de la distribución normal de tu muestra
mu = datos_diferencia.mean()
sigma = datos_diferencia.std()

# 3. Realizar el Test de Kolmogorov-Smirnov
# Comparamos 'datos_diferencia' contra la distribución normal ('norm') 
# pasándole la media (loc) y desviación estándar (scale) calculadas.
d_stat, p_value = stats.kstest(datos_diferencia, 'norm', args=(mu, sigma))

# 4. Mostrar e interpretar resultados
print("--- TEST DE NORMALIDAD (KOLMOGOROV-SMIRNOV) ---")
print(f"Estadístico D: {d_stat:.4f}")
print(f"P-valor: {p_value:.4f}")

print("\nInterpretación:")
alpha = 0.05 # Nivel de significancia común
if p_value > alpha:
    print("No se puede rechazar la hipótesis nula (H0).")
    print("Los datos parecen seguir una distribución NORMAL.")
else:
    print("Se rechaza la hipótesis nula (H0).")
    print("Los datos NO siguen una distribución normal.")
```

### 1.2) Test t Student para la totalidad de las muestras metodo de extraccion por tierras

```python
# 1. Limpieza de datos
# Es fundamental eliminar filas con valores nulos en cualquiera de las dos columnas
# para que las parejas de datos coincidan perfectamente.
df_limpio = df3[['ext_cut_rip_ph', 'ext_cut_rip_selec_col']].dropna()

# 2. Realizar el Test T de Student para muestras relacionadas
t_stat, p_value = stats.ttest_rel(df_limpio['ext_cut_rip_ph'], df_limpio['ext_cut_rip_selec_col'])

# 3. Mostrar e interpretar resultados
print("--- TEST T DE STUDENT (MUESTRAS RELACIONADAS) ---")
print(f"Estadístico t: {t_stat:.4f}")
print(f"P-valor: {p_value:.10f}")

print("\nInterpretación:")
alpha = 0.05 # Nivel de significancia
if p_value < alpha:
    print("Resultado: SIGNIFICATIVO.")
    print("Hay una diferencia estadísticamente significativa entre la extracción Iso pH y la de Columna.")
    
    # Verificamos cuál es mayor
    media_ph = df_limpio['ext_cut_rip_ph'].mean()
    media_col = df_limpio['ext_cut_rip_selec_col'].mean()
    if media_ph > media_col:
        print(f"En promedio, la extracción Iso pH ({media_ph:.2f}) es superior a la de Columna ({media_col:.2f}).")
    else:
        print(f"En promedio, la extracción por Columna ({media_col:.2f}) es superior a la de Iso pH ({media_ph:.2f}).")
else:
    print("Resultado: NO SIGNIFICATIVO.")
    print("No hay evidencia suficiente para afirmar que un método sea diferente al otro.")
```

```python
# Confirmación de test t Student
# =========================
t_stat, t_p = ttest_rel(df3["ext_cut_rip_selec_col"], df3["ext_cut_rip_ph"], alternative="two-sided")
print("T pareado: t =", t_stat, "p =", t_p)
```

## 2) Test de hipótesis para extracción de cobre mediante cabeza analizada

### 2.1) Test de hipotesis normalidad para la totalidad de las muestras metodo cabeza analizada

```python
import pandas as pd
from scipy import stats
from scipy.stats import skew, kurtosis

# 1. Definición de columnas y carga de datos (Análisis de Solución)
columnas_deseadas = ['ext_cut_anlz_ph', 'ext_cut_anlz_col']
excelPath = r"C:\Users\Diego\Desktop\Ultimo Esfuerzo\Copia de Base de Datos Cuprochlor Sin T Final.xlsx"

# Creamos el nuevo DataFrame df4
df4 = pd.read_excel(excelPath, "BD elim y reducido (2)", usecols=columnas_deseadas)

# 2. Cálculo de la variable diferencia en df4
df4['diferencia'] = df4['ext_cut_anlz_ph'] - df4['ext_cut_anlz_col']

# 3. Limpieza de datos nulos
datos_diferencia = df4['diferencia'].dropna()

# 4. Cálculo de Asimetría y Curtosis sobre df4
asimetria = skew(datos_diferencia)
curto = kurtosis(datos_diferencia)

print(f"--- ESTADÍSTICOS DE FORMA (TOTAL MUESTRAS - DF4) ---")
print(f"Asimetría: {asimetria:.4f}")
print(f"Curtosis:  {curto:.4f}\n")

# 5. Parámetros para el Test de Kolmogorov-Smirnov
mu = datos_diferencia.mean()
sigma = datos_diferencia.std()

# 6. Realización del Test de Kolmogorov-Smirnov
d_stat, p_value = stats.kstest(datos_diferencia, 'norm', args=(mu, sigma))

# 7. Mostrar e interpretar resultados
print("--- TEST DE NORMALIDAD (KOLMOGOROV-SMIRNOV) ---")
print(f"Estadístico D: {d_stat:.4f}")
print(f"P-valor: {p_value:.4f}")

print("\nInterpretación:")
alpha = 0.05 
if p_value > alpha:
    print("Resultado: No se puede rechazar la hipótesis nula (H0).")
    print("Los datos en df4 parecen seguir una distribución NORMAL.")
else:
    print("Resultado: Se rechaza la hipótesis nula (H0).")
    print("Los datos en df4 NO siguen una distribución normal.")

print(f"\nCantidad de muestras analizadas en df4: {len(datos_diferencia)}")
```

### 2.2) Test t Student para la extracción de cobre calculada mediante método de cabeza analizada

```python
import pandas as pd
from scipy import stats

# 1. Preparación y Limpieza de datos en df4
# Seleccionamos las variables de análisis de solución y eliminamos nulos para asegurar pares completos
df_limpio = df4[['ext_cut_anlz_ph', 'ext_cut_anlz_col']].dropna()

# 2. Realizar el Test T de Student para muestras relacionadas
# Comparamos la extracción analizada por solución: pH vs Columna
t_stat, p_value = stats.ttest_rel(df_limpio['ext_cut_anlz_ph'], df_limpio['ext_cut_anlz_col'])

# 3. Mostrar e interpretar resultados
print("--- TEST T DE STUDENT (MUESTRAS RELACIONADAS - ANÁLISIS SOLUCIÓN) ---")
print(f"Estadístico t: {t_stat:.4f}")
print(f"P-valor: {p_value:.10f}")
print(f"Cantidad de muestras analizadas en df4: {len(df_limpio)}")

print("\nInterpretación:")
alpha = 0.05  # Nivel de significancia
if p_value < alpha:
    print("Resultado: SIGNIFICATIVO.")
    print("Hay una diferencia estadísticamente significativa entre la extracción Iso pH y la de Columna (vía solución).")
    
    # Cálculo de medias para determinar el método superior en df4
    media_ph = df_limpio['ext_cut_anlz_ph'].mean()
    media_col = df_limpio['ext_cut_anlz_col'].mean()
    
    if media_ph > media_col:
        print(f"En promedio, la extracción Iso pH ({media_ph:.2f}%) es significativamente superior a la de Columna ({media_col:.2f}%).")
    else:
        print(f"En promedio, la extracción por Columna ({media_col:.2f}%) es significativamente superior a la de Iso pH ({media_ph:.2f}%).")
else:
    print("Resultado: NO SIGNIFICATIVO.")
    print("No hay evidencia suficiente para afirmar que un método sea diferente al otro en el análisis de solución.")
```
